```python
# import pprint  # Pretty print data structures
import numpy as np
import pandas as pd
import math
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter
plt.style.use('ggplot')
```

```python
def plot_distribution(df, label, save=None, format='pdf', xlim=(1.5, 12)):
    plt.figure(figsize=(6, 3.5))
    plt.rcParams.update({
        'text.usetex': True,
        'text.latex.preamble': r'\usepackage{amsmath}',
        'font.family': 'serif',
        'font.serif': ['Computer Modern Roman'],
        'mathtext.fontset': 'cm',
        })
    bin_width = 0.25
    bin_train = np.arange(min(df), max(df), bin_width)
    plt.hist(df, bins=bin_train, label=rf'$N_{{\mathrm{{tot}}}}$ = {df.shape[0]:,}')
    plt.xlabel(label, fontsize=16, color='black')
    plt.ylabel('Count', fontsize=16, color='black')
    plt.xlim(*xlim)
    plt.tick_params(axis='both', labelsize=14, colors='black')
    plt.gca().xaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))
    plt.gca().yaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))
    plt.legend(facecolor='white', fontsize=14)
    plt.tight_layout()
    if save is not None:
        plt.savefig(save, format=format, dpi=300,)
    plt.show()
    print(max(df))
    print(min(df))
```

```python
df_qm9 = pd.read_csv('qm9/qm9.csv')['gap']*27.2114
df_qm9small = pd.read_csv('qm9/qm9small.csv')['gap']*27.2114
df_matbench = pd.read_csv('matbench/e_form/matbench_composition.csv')['e_form']
df_hse = pd.read_csv('hse_df_feat/hse_df_feat.csv')['dG_hse06 (eV/atom)']
```

```python
plot_distribution(df_qm9, r'$\Delta E_{gap}$ (eV)', xlim=(0.5,12.5), save='qm9/gap/qm9_distribution.pdf')
```

```python
plot_distribution(df_qm9small, r'$\Delta E_{gap}$ (eV)', xlim=(2,11), save='qm9/gap/qm9small_distribution.pdf')
```

```python
plot_distribution(df_matbench, r'$\Delta E_{f}$ (eV/atom)', xlim=(-5, 3), save='matbench/e_form/matbench_distribution.pdf')
```

```python
plot_distribution(df_hse, r'$\Delta G_{pbx}$ (eV/atom)', xlim=(-1.5, 13), save='hse_df_feat/dg/hse_distribution.pdf')
```
